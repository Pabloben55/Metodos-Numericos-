#Método de Newton


En este método vamos a encontrar una aproximación a la raíz de una función (f(x)=0), partiendo de un valor inicial $(p_0)$. Para calcular las siguientes aproximaciones, utilizamos la fórmula de Newton:

$[p_n=p_{n-1}-\frac{f(p_{n-1})}{f'(p_{n-1})}]$

Repetimos este procedimiento hasta que la diferencia entre dos aproximaciones consecutivas sea menor que la tolerancia establecida, es decir,

$[|p_n-p_{n-1}|<\text{TOL}]$

Si se cumple esta condición, obtenemos una solución aproximada. Si se alcanza el número máximo de iteraciones sin cumplirla, el método termina e indica que no se encontró una solución con la precisión deseada.

<h3>
<Font color='pink',font-family='verdana'>
Cálculo de aproximaciones  
</h3>
</font>

```python
def newton(p0,tol):
  error=100
  Lerror=[]
  Lp_i=[p0]
  while error>tol:
    p1=p0-(np.cos(p0) - p0)/(-np.sin(p0)-1)
    error=np.abs(p1-p0)
    Lerror.append(error)
    Lp_i.append(p1)
    p0=p1
  return Lp_i,Lerror
```

```python
p = newton(0.74, 0.00001)

tabla = pd.DataFrame({
    "n": range(1, len(p[1]) + 1),
    "p_n-1": p[0][:-1],
    "p_n": p[0][1:],
    "|p_n - p_n-1|": p[1]
})
#hacemos la tabla respecto a las iteraciones
print(tabla)
```
