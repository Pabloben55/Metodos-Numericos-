#Metodo del Punto Fijo

El método de punto fijo busca una solución de la forma

$$ x=g(x) $$

Se comienza con una aproximación inicial (p_0) y se van calculando nuevos valores mediante

$$ p=g(p_0). $$

El proceso continúa hasta que la diferencia entre dos valores consecutivos sea menor que la tolerancia:

$$ |p-p_0|<TOL. $$

De esta manera, se obtiene una aproximación al punto fijo y, por lo tanto, a la solución de la ecuación.

Usaremos el ejemplo la siguiente ecuación:

$$ f(x)=x^3+4x^2-10=0 $$

Primero debemos escribirla en la forma

$$ x=g(x) $$

y después elegir una aproximación inicial (p_0). En cada iteración calcularemos un nuevo valor de (p) hasta que el error sea menor que

$$ 10^{-4}. $$

El objetivo es encontrar una aproximación de la raíz con una precisión de al menos 10
−4 y para esto utilizaremos una raiz que sea continua en $[1,2]$

##Solución

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
#Importamos pandas para poder hacer la tabla posteriormente
```

<b>
<font color="Green" face="Verdana">
Definimos la Ecuación
</font>
</b>



```python
def f(x):
  return x**3+4*x**2-10 #definimos la funcion
```

###Generamos la Gráfica de la función

```python
xx=np.linspace(1,2)
plt.plot(xx,f(xx),color='red')
plt.axhline(y=0,color='black')
plt.axvline(x=0,color='black')
#Generamos la grafica
```

####Ahora usamos la funcion $g1(x)$

```python
def g1(x):
  return x-((x**3+4*x**2-10)/(3*x**2+8*x)) #definimos la funcion autilizar
```

<H2>
<font color="Pink" face="Times New Roman">
Evaluamos para verificar que hay punto fijo  en $[1,2]$
</font>
<h2>

```python
g1(1)
```

```python
g1(2)
```

<H3>
<font color="Violet" face="Palatino">
Definimos la aproximacion inicial como:  $p0=1.5$
</font>
<H3>

```python
def p_fijo(p0, i):

    lista_p = [] #guarda el valor anterior de  p_n que es el valor qu le dimos
    lista_g = []#guarda p1 que es el calculo de g(p_n)
    error = [] #guarda |P_n-P_n-1|

    for n in range(i):
        p1 = g1(p0)

        lista_p.append(p0)
        lista_g.append(p1)

        e = np.abs(p1-p0)
        error.append(e)

        p0 = p1
 #El nuevo valor se usa en la siguiente iteración

    return lista_p, lista_g, error

p = p_fijo(1.5, 5)

tabla = pd.DataFrame({ "n": range(1, len(p[0]) + 1), "p_n": p[0], "g(p_n)": p[1], "|P_n-P_n-1|": p[2] })
print(tabla)
```

<h2>
<font color="RED", face="verdana">
Se aproximan cada vez más a $(1.3652)$. Por lo tanto, el método converge al punto fijo $(papprox=1.3652)$.
</font>
</h2>
