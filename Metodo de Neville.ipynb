<H1> METODO DE NEVILLE



<h4> Veámoslo con un $EJEMPLO$:

```python
import numpy as np #importamos librerias
import matplotlib.pyplot as plt
import sympy as sp
```


Para evaluar el polinomio de interpolación $P$ en los diferentes números $n + 1$, $x_0$, ... , $x_n$ en
el número $x$ para la función $f$ :



```python
X = [1.0, 2,3,4,5] #datos de los n+1 nodos arbitrarios
Y= [1,4,9,16,25]
```

Se desarrolla una función que permita calcular los polinomios de interpolación de distintos grados, comenzando desde el $grado 0, grado 1, grado 2$ y continuando sucesivamente.

```python
def metodo_neville(X,Y, x_evaluar=None):
    n = len(X)

    if x_evaluar is None:
        x = sp.Symbol('x')
        Q = sp.zeros(n, n)
    else:
        x = x_evaluar
        Q = np.zeros((n, n))
#La matriz numerica
    for i in range(n):
        Q[i, 0] = Y[i]

    for j in range(1, n):

        for i in range(j, n):
  # fórmula de Neville
            termn1 = (x - X[i-j]) * Q[i, j-1]
            termn2 = (x - X[i]) * Q[i-1, j-1]
            denominador = X[i] - X[i-j]

            Q[i, j] = (termn1 - termn2) / denominador

    polinomio_final = Q[n-1, n-1]

    return polinomio_final, Q


print(" Calculamos el polinomio simbólico")
polinomio, tabla_Q = metodo_neville(X, Y)


polinomio_simplificado = sp.simplify(sp.expand(polinomio))
polinomio_redondeado = sp.N(polinomio_simplificado, 6)

print("El polinomio p(x) que pasa por los nodos es:")
print(f"P(x) = {polinomio_redondeado}")


```

```python
x_punto = 1.5
print(f"\n--- Evaluamos en x = {x_punto} ---")
resultado_evaluado, _ = metodo_neville(X1,Y1, x_evaluar=x_punto)
print(f"p({x_punto}) = {resultado_evaluado:.8f}")

_, tabla_numerica = metodo_neville(X1, Y1, x_evaluar=x_punto)
n = len(X)

for j in range(n):
    print(f"Polinomios del grado {j}:")
    for i in range(j, n):
        # Para que esten alineados ponemos
        print(f"  Q_{i},{j} = {tabla_numerica[i, j]:.8f}")
    print("-" * 40)


```
