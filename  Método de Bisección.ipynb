#Método de Bisección


A continuación, aplicaremos el **método de bisección**, un método numérico utilizado para aproximar las raíces de una función. Para comprender su procedimiento, utilizaremos como ejemplo la siguiente ecuación:

$$
x^3+4x^2-10=0
$$

considerando el intervalo

$$
[1,2].
$$

Para aplicar este método de bisección debemos encontrar una solucion
$$
f(x) = 0
$$ que dada la función continua determinada f en el intervalo $[a, b]$, donde $f(a)$ y $f(b)$  tienen signos opuestos:


$$
f(a)\cdot f(b)<0.
$$

Si esta condición se cumple, por el **Teorema del Valor Intermedio** podemos asegurar que existe al menos una raíz de la función dentro del intervalo.

A partir de esto, iremos aproximando la raíz de manera **iterativa**, dividiendo el intervalo en dos partes y seleccionando en cada paso el subintervalo que contiene la raíz.

Para ello, calcularemos el **punto medio** de cada intervalo mediante la fórmula

$$
p_i=\frac{a_i+b_i}{2}.
$$

y posteriormente se evalúa $(f(p))$. Dependiendo de los signos de $(f(a)$) y $(f(p)$), se selecciona el subintervalo que contiene la raíz y se repite el procedimiento.

El proceso continúa hasta encontrar la raíz con la precisión deseada, considerando como criterio de paro que

$$ f(p)=0 $$

o que

$$ \frac{b-a}{2}<TOL, $$

donde $(TOL)$ representa la tolerancia o error máximo permitido(Número maximo de Iteraciones).


```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

###Valores

<b>
<font color="blue" face="Verdana">
Guardamos los valores iniciales
</font>
</b>



```python
a=1
b=2
```

<b>
<font color="blue" face="Verdana">
Guardamos los valores iniciales
</font>
</b>

```python
a0=a
b0=b
```

<b>
<font color="blue" face="Verdana">
Guardamos los valores iniciales del error
</font>
</b>

```python
tol=0.000001
nmax=100
error=100
niter=0
```

###Método de Bisección

####Resolución

<b>
<font color="green" face="Verdana">
Evaluamos el  primer valor medio
</font>
</b>

```python
m = a + (b - a)/2
```

<b>
<font color="green" face="Verdana">
Evaluamos la función en los puntos a, b y m
</font>
</b>

```python
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error ))
```

<b>
<font color="green" face="Verdana">
CICLO ITERATIVO
</font>
</b>

```python
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = f(a)
    else:
        b = m
        fb = f(b)

    m = a + (b - a)/2
    fm = f(m)
    error = abs(b - a)
    niter += 1
```

##RESULTADO

```python
print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```
